# Reinforcement Learning

Practical revision notebook:

**Basic workflow → major RL algorithms → compare → model improvement**

Reinforcement Learning (RL) learns by interaction:

**Agent → Action → Environment → Reward → New State → Learn**

Unlike supervised learning, the model is not given the correct answer for every step.

## Basic Workflow

1. Define environment
2. Define states
3. Define actions
4. Define rewards
5. Agent chooses action
6. Environment returns next state + reward
7. Update policy/value
8. Repeat for episodes
9. Evaluate policy
10. Improve hyperparameters

## Basic RL Terms

- **Agent** → learner/decision maker
- **Environment** → world the agent interacts with
- **State** → current situation
- **Action** → choice made by the agent
- **Reward** → feedback from the environment
- **Policy** → strategy for choosing actions
- **Episode** → one complete run
- **Value** → expected future reward

# Q-Learning

**Use:** Small/discrete state and action spaces where you can store a Q-table.

Main idea:

**Q(state, action) → expected future reward**

The agent learns which action is valuable in each state.

```python
# Q-Learning core update

Q[state, action] = Q[state, action] + alpha * (
    reward
    + gamma * max(Q[next_state])
    - Q[state, action]
)

# alpha = learning rate
# gamma = discount factor
```

# SARSA

**Use:** Discrete state/action problems when you want an on-policy learning method.

Main difference:

**Q-Learning → learns toward the best next action**

**SARSA → learns using the action actually selected by the current policy**

```python
# SARSA update

Q[state, action] = Q[state, action] + alpha * (
    reward
    + gamma * Q[next_state, next_action]
    - Q[state, action]
)
```

## Model Comparison

| Algorithm | Main idea | Best use |
|---|---|---|
| Q-Learning | Q-table, off-policy | Small discrete problems |
| SARSA | On-policy Q-learning | Discrete problems |
| DQN | Neural network estimates Q-values | Large/discrete state spaces |
| Policy Gradient | Learn policy directly | Complex policies |
| Actor-Critic | Actor + value critic | Larger/continuous control |
| PPO | Stable clipped policy updates | General modern RL/control |

### Quick selection

**Small discrete problem → Q-Learning / SARSA**

**Large discrete state space → DQN**

**Complex or continuous control → Actor-Critic / PPO**

## Model Improvement

**Baseline → Tune exploration → Tune learning parameters → Improve reward design → Train for more episodes → Evaluate policy → Final policy**

### Q-Learning / SARSA
Tune:
- `alpha` → learning rate
- `gamma` → discount factor
- `epsilon` → exploration
- epsilon decay

### DQN
Tune:
- Learning rate
- Batch size
- Replay buffer size
- Discount factor
- Exploration schedule
- Target-network update frequency

### PPO
Tune:
- Learning rate
- Clip range
- Batch size
- Number of epochs
- Discount factor
- Entropy coefficient

### Important

RL performance is strongly affected by:
- Reward design
- Exploration strategy
- Environment
- Randomness

Do not judge an RL agent from one training run.
