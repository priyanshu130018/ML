```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from src.data.loader import ChurnDataLoader

plt.style.use("ggplot")
```

# Missing Values

```python
df.isnull().sum()
```

# Target Variable Distribution

```python
df["Churn"].value_counts()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Churn"
)

plt.title("Customer Churn Distribution")

plt.show()
```

# Numerical Features

```python
numerical_columns = df.select_dtypes(
    include=np.number
).columns

numerical_columns
```

```python
df[numerical_columns].describe()
```

# Distribution of Numerical Features

```python
df[numerical_columns].hist(
    figsize=(12,8),
    bins=20
)
plt.tight_layout()
plt.show()
```

# Correlation Heatmap

```python
plt.figure(figsize=(8,6))
sns.heatmap(
    df[numerical_columns].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Matrix")
plt.show()
```

# Categorical Features

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

categorical_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns

categorical_columns
```

# Categorical Features

```python
categorical_columns = categorical_columns.drop("customerID")

for col in categorical_columns:
    plt.figure(figsize=(6,4))
    sns.countplot(data=df, x=col)
    plt.xticks(rotation=45)
    plt.title(col)
    plt.tight_layout()
    plt.show()
```

# Churn vs Gender

```python
plt.figure(figsize=(6,4))
sns.countplot(
    data=df,
    x="gender",
    hue="Churn"
)
plt.show()
```

# Churn vs Contract Type

```python
plt.figure(figsize=(7,4))
sns.countplot(
    data=df,
    x="Contract",
    hue="Churn"
)
plt.show()
```

# Churn vs Internet Service

```python
plt.figure(figsize=(7,4))
sns.countplot(
    data=df,
    x="InternetService",
    hue="Churn"
)
plt.show()
```

# Churn vs Payment Method

```python
plt.figure(figsize=(10,4))
sns.countplot(
    data=df,
    x="PaymentMethod",
    hue="Churn"
)
plt.xticks(rotation=20)
plt.show()
```

# Monthly Charges Distribution

```python
plt.figure(figsize=(7,4))
sns.histplot(
    data=df,
    x="MonthlyCharges",
    bins=30
)
plt.show()
```

# Monthly Charges vs Churn

```python
plt.figure(figsize=(7,4))
sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)
plt.show()
```

# Tenure vs Churn

```python
plt.figure(figsize=(7,4))
sns.boxplot(
    data=df,
    x="Churn",
    y="tenure"
)
plt.show()
```

# Pairplot

```python
sns.pairplot(
    df[
        [
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "Churn"
        ]
    ],
    hue="Churn"
)
```

# Conclusion

- The dataset contains no missing values or duplicate records.
- Most customers do not churn, indicating a moderately imbalanced target variable.
- Contract type, tenure, monthly charges, and internet service show strong relationships with customer churn.
- Customers with month-to-month contracts are more likely to churn.
- Customers with longer tenure are more likely to remain with the company.
- Categorical features require encoding before model training.
- Numerical features will be standardized using `StandardScaler` before training the Logistic Regression model.
